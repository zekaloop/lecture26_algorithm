```python
from collections import deque
queue = deque()

def enqueue(data):
    queue.append(data)

def dequeue():
    if queue:
        data = queue.popleft()
    else:
        data = -1
    return data

def getFront():
    if queue: return queue[0]
    return -1
    
while True:
    #메뉴 프린트

    print('===========================================')
    print('1.Enqueue  2. Dequeue  3.Get Front  0.Exit')
    print('===========================================')
    #메뉴 번호 입력
    menu = int(input('>> 연산 메뉴 선택'))

    
    if menu == 0: break

    elif menu == 1: # enqueue
        data = int(input('데이터 :'))
        enqueue(data)
        

    elif menu == 2: # deque
        data = dequeue()
        if data != -1:
            print('가져온 데이터:', data)
        else:
            print('Queue가 비어있습니다')

    elif menu == 3: # peek
        data = getFront()
        if data != -1:
            print(data)
        else:
            print('Queue가 비어있습니다')

    print(queue)
        

        
print('종료')
```
