```python
# insertion sort
def insertion_sort(A):
    for cur_index in range(1, len(A)):
        #해당 데이터의 위치를 찾아서 insertion
        key = A[cur_index]
        i = cur_index - 1
        while i >= 0 and A[i] > key:
            A[i+1] = A[i]
            i -= 1

        A[i+1] = key
        print(cur_index, A)
```

```python
def bubble_sort(A):
    for last in range(len(A)-1, 0, -1):
        is_changed = False
        for i in range(last):
            if A[i] > A[i+1]:
                A[i], A[i+1] = A[i+1], A[i]
                is_change = True
        print(last, A)
        if not is_changed: return
```

```python
A = [23, 5, 44, 1, 6]
print('원래 데이터 : ', A)

bubble_sort(A)
print('정렬 후 데이터: ', A)
```
